## Dimensions in Arrays

Create arrays of different dimentions.

a=A numpy array with one single integer 10

b=A numpy array passing a list having a list= [1,2,3]

c=A numpy array passing nested list having [[1, 2, 3], [4, 5, 6]] as elements

d=A numpy array passing nested list having [[[1, 2, 3], [4, 5, 6]], [[1, 2, 3], [4, 5, 6]]] as elements

```python
import numpy as np

# a: 0-D array (scalar)
a = np.array(10)

# b: 1-D array
b = np.array([1, 2, 3])

# c: 2-D array
c = np.array([[1, 2, 3], [4, 5, 6]])

# d: 3-D array
d = np.array([[[1, 2, 3], [4, 5, 6]], 
              [[1, 2, 3], [4, 5, 6]]])

# Printing arrays
print("a =", a)
print("b =", b)
print("c =\n", c)
print("d =\n", d)

# Checking dimensions
print("Dimension of a:", a.ndim)
print("Dimension of b:", b.ndim)
print("Dimension of c:", c.ndim)
print("Dimension of d:", d.ndim)
```

Are you ready to check its dimention? Use ndim attribute on each variable to check its dimention

```python
#print dimentions of a,b, c and d


import numpy as np

a = np.array(10)
b = np.array([1, 2, 3])
c = np.array([[1, 2, 3], [4, 5, 6]])
d = np.array([[[1, 2, 3], [4, 5, 6]], 
              [[1, 2, 3], [4, 5, 6]]])

print(a.ndim)  # Output: 0
print(b.ndim)  # Output: 1
print(c.ndim)  # Output: 2
print(d.ndim)  # Output: 3
```

Hey It's True they both share memory

Woh! simple assignment is similar to view. That means

**hstack vs vstack function**


Stacking is same as concatenation, the only difference is that stacking is done along a new axis.

NumPy provides a helper function: 

1. hstack() to stack along rows.
2. vstack()  to stack along columns

```python
import numpy as np

result = np.hstack((x1, x4))

print(result)
```

```python
import numpy as np

result = np.vstack((x1, x4))

print(result)
```

We hope now you saw the difference between them.

Fun fact! you can even use concatenate() function to join 2 arrays along with the axis. If axis is not explicitly passed, it is taken as 0 ie. along column

Lets try this function as well

```python
import numpy as np

arr1 = np.array([[1, 2],
                 [3, 4]])

arr2 = np.array([[5, 6],
                 [7, 8]])

result = np.concatenate((arr1, arr2), axis=0)

print(result)
```

```python
import numpy as np

arr1 = np.array([[1, 2],
                 [3, 4]])

arr2 = np.array([[5, 6],
                 [7, 8]])

result = np.concatenate((arr1, arr2), axis=1)

print(result)
```

## Adding, Insert and delete Numpy array

You can also add 2 arrays using append() function also. This function appends values to end of array

Lets see how

```python
import numpy as np

arr1 = np.array([[1, 2],
                 [3, 4]])

arr2 = np.array([[5, 6],
                 [7, 8]])

result = np.append(arr1, arr2)

print(result)
```

Lets use insert() function which Inserts values into array before specified index value

```python
import numpy as np

result = np.insert(x1, 4, x4)

print(result)
```

You can see in above output we have inserted all the elements of x4 before index 4 in array x1.

```python
import numpy as np

result = np.delete(x2, 1)

print(result)
```
